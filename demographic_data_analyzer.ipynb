# Demographic Data Analyzer

In this challenge you must analyze demographic data using Pandas. You are given a dataset of demographic data that was extracted from the 1994 Census database. 


You must use Pandas to answer the following questions:

- How many people of each race are represented in this dataset? This should be a Pandas series with race names as the index labels. (race column)
- What is the average age of men?
- What is the percentage of people who have a Bachelor's degree?
- What percentage of people with advanced education (Bachelors, Masters, or Doctorate) make more than 50K?
- What percentage of people without advanced education make more than 50K?
- What is the minimum number of hours a person works per week?
- What percentage of the people who work the minimum number of hours per week have a salary of more than 50K?
- What country has the highest percentage of people that earn >50K and what is that percentage?
- Identify the most popular occupation for those who earn >50K in India.

```python
import pandas as pd
df = pd.read_csv('adult.data.csv')
df.head(3)
```

Q1. How many people of each race are represented in this dataset? This should be a Pandas series with race names as the index labels. (race column)

```python
ppl_of_each_race = df['race'].value_counts()
print(f'Total no. of people of each race are :\n {ppl_of_each_race}')
```

Q2. What is the average age of men?

```python
avg_mens_age = df[df['sex'] == 'Male']['age'].mean()

print(f'Average Age of Men is {round(avg_mens_age, 1)}')
```

Q3. What is the percentage of people who have a Bachelor's degree?

```python
bachelors_count = (df['education'] == 'Bachelors').sum()
total_people = len(df)
percentage_bachelors = (bachelors_count / total_people) * 100

print(f"Percentage of people with a Bachelor's degree: {round(percentage_bachelors, 1)}%")
```

Q4. What percentage of people with advanced education (Bachelors, Masters, or Doctorate) make more than 50K?

```python
edu = ['Bachelors', 'Masters', 'Doctorate']

adv_edu = df[df["education"].isin(edu)]
per_of_ppl = (adv_edu['salary'] == '>50K').mean()*100
print(f"Percentage of people with advanced education making >50K: {round(per_of_ppl, 1)}%")
```

Q5. What percentage of people without advanced education make more than 50K?

```python
edu = ['Bachelors', 'Masters', 'Doctorate']

adv_edu = df[~df["education"].isin(edu)]
per_of_ppl = (adv_edu['salary'] == '>50K').mean()*100
print(f"Percentage of people with advanced education making >50K: {round(per_of_ppl, 1)}%")
```

Q6. What is the minimum number of hours a person works per week?

```python
min_work_hours = df['hours-per-week'].min()
print(f"The minimum number of hours a person works per week is {min_work_hours} hour.")
```

Q7. What percentage of the people who work the minimum number of hours per week have a salary of more than 50K?

```python
min_workers = df[df['hours-per-week'] == min_work_hours]
rich_percentage = (min_workers['salary'] == '>50K').mean() * 100

print(f"Percentage of minimum hour workers making >50K: {rich_percentage}%")
```

Q8. What country has the highest percentage of people that earn >50K and what is that percentage?

```python
country_total = df['native-country'].value_counts()
country_rich = df[df['salary'] == '>50K']['native-country'].value_counts()

country_percentages = (country_rich / country_total) * 100

highest_earning_country = country_percentages.sort_values(ascending=False)

print(highest_earning_country.head(1))
```

Q9. Identify the most popular occupation for those who earn >50K in India.

```python
sal = df[(df['salary'] == '>50K') & (df['native-country'] == 'India')]
occupation_count = sal["occupation"].value_counts()
occupation_count.head(1)
```
